# Prime Number — Mathematics & Python Logic

This notebook explains the mathematics behind checking whether a number is prime, especially why we only check divisors up to `√num` and how the Python line

```python
for i in range(2, int(num**0.5) + 1):
```

works.

## 1. What is a Prime Number?

A **prime number** is a number greater than `1` that has exactly two positive factors:

- `1`
- The number itself

Examples:

- `2 → 1 × 2` → Prime
- `3 → 1 × 3` → Prime
- `5 → 1 × 5` → Prime
- `7 → 1 × 7` → Prime

A number such as `6` is not prime because:

`6 = 1 × 6 = 2 × 3`

So `6` has factors other than `1` and itself.

## 2. The Basic Mathematical Idea

To determine whether `num` is prime, we try to find a divisor other than `1` and `num`.

If we find even one divisor:

`num % i == 0`

then the number is **not prime**.

If no divisor is found, the number is **prime**.

## 3. Why Do We Only Check Up to √num?

This is the most important mathematical idea in the program.

Consider `num = 36`.

Its factor pairs are:

```text
1 × 36
2 × 18
3 × 12
4 × 9
6 × 6
9 × 4
12 × 3
18 × 2
36 × 1
```

The factor pairs meet at:

`6 × 6 = 36`

And:

`√36 = 6`

After `√36`, we are only seeing the other side of the same factor pairs. Therefore, checking beyond `√num` is unnecessary.

## 4. The Mathematical Rule

For any integer `N > 1`:

> If `N` is not prime, it must have at least one factor less than or equal to `√N`.

Suppose:

`N = a × b`

If both `a` and `b` were greater than `√N`, then:

`a > √N` and `b > √N`

Multiplying both inequalities gives:

`a × b > √N × √N`

Therefore:

`a × b > N`

But we already know:

`a × b = N`

This is impossible. Therefore, at least one factor must be `≤ √N`.

## 5. Understanding `num**0.5`

In Python:

```python
num**0.5
```

means:

`num^0.5 = √num`

Example:

```python
36**0.5
```

gives:

`6.0`

So `num**0.5` is simply another way of calculating the square root.

````python
num = 36
print(num**0.5)
````

## 6. Why Do We Use `int()`?

The square root may be a decimal.

For example:

`√29 ≈ 5.385`

Python gives approximately `5.385...`.

Using:

```python
int(num**0.5)
```

converts it to the integer part:

`int(5.385...) = 5`

We only need the integer boundary for `range()`.

## 7. Why Do We Use `+ 1`?

Python's `range(start, stop)` does **not include the stop value**.

For example:

```python
range(2, 6)
```

produces:

`2, 3, 4, 5`

It does not include `6`.

For `36`, we need to check through `√36 = 6`.

Therefore:

```python
range(2, int(36**0.5) + 1)
```

becomes:

```python
range(2, 7)
```

which gives:

`2, 3, 4, 5, 6`

The `+1` makes the square-root boundary inclusive.

## 8. Understanding the Important Line

The line is:

```python
for i in range(2, int(num**0.5) + 1):
```

Break it into four pieces:

### `for i`
We create a variable `i` that represents a possible divisor.

### `range(2, ...)`
We start checking from `2` because `1` divides every number and therefore cannot prove that a number is composite.

### `num**0.5`
This calculates `√num`.

### `+ 1`
Python excludes the final value in `range()`, so we add `1` to include `√num` when it is an integer.

In plain English:

> Check every possible divisor from `2` up to `√num`.

## 9. Understanding `%`

The `%` operator gives the **remainder** after division.

For example:

`10 % 2 = 0`

because `10 ÷ 2` has remainder `0`.

But:

`10 % 3 = 1`

because `10 ÷ 3` leaves remainder `1`.

Therefore:

```python
num % i == 0
```

means:

> `i` divides `num` exactly.

## 10. Example: Checking 29

First calculate:

`√29 ≈ 5.385`

So we check:

`2, 3, 4, 5`

Now calculate:

```text
29 % 2 = 1
29 % 3 = 2
29 % 4 = 1
29 % 5 = 4
```

None of the results is `0`.

Therefore, no divisor was found and `29` is prime.

## 11. Example: Checking 35

First:

`√35 ≈ 5.91`

So check:

`2, 3, 4, 5`

The results are:

```text
35 % 2 = 1
35 % 3 = 2
35 % 4 = 3
35 % 5 = 0
```

We found a divisor: `5`.

Therefore:

`35 = 5 × 7`

So `35` is not prime.

Notice that `7 > √35`. We did not need to check `7` because finding `5` was already enough.

## 12. The Important Mistake in the Original Code

The original code has this structure:

```python
if num % i == 0:
    print(f"{num} is not a prime number")
    break
else:
    print(f"{num} is a prime number")
```

The problem is that the `else` belongs to the `if`.

If one divisor does not work, the code immediately prints `prime` even though other possible divisors still need to be checked.

For example, for `7`:

```text
7 % 2 != 0 → prints prime
7 % 3 != 0 → prints prime again
7 % 4 != 0 → prints prime again
7 % 5 != 0 → prints prime again
```

That is logically incorrect because we should only conclude that the number is prime **after all possible divisors have been checked**.

## 13. Correct Prime Number Program

The clean solution is to use Python's `for-else` structure.

````python
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            print(f"{num} is not a prime number")
            break
    else:
        print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
````

## 14. How `for-else` Works Here

This is a Python-specific concept.

```python
for i in range(...):
    if condition:
        break
else:
    ...
```

The `else` executes only when the loop finishes normally, meaning **no `break` happened**.

So the logic becomes:

```text
Found a divisor
      ↓
     break
      ↓
NOT PRIME
```

or:
```text
Checked every possible divisor
      ↓
No divisor found
      ↓
Loop finishes without break
      ↓
PRIME
```

## 15. Complete Logic in Simple Steps

For a number `num`:

1. Check whether `num > 1`.
2. Calculate `√num`.
3. Start checking possible divisors from `2`.
4. Stop at `√num`.
5. For each divisor `i`, calculate `num % i`.
6. If the remainder is `0`, a divisor was found → **not prime**.
7. Use `break` because one divisor is enough to prove the number is not prime.
8. If the entire loop finishes without `break`, no divisor exists → **prime**.

## 16. Mental Model to Remember

Think of the algorithm as:

```text
Is num greater than 1?
        ↓
   Find √num
        ↓
Check 2, 3, 4, ... √num
        ↓
Does any number divide num exactly?
      /       \
    YES        NO
     ↓          ↓
 NOT PRIME    PRIME
```

### One sentence to remember

**To determine whether `N` is prime, you only need to search for a divisor from `2` through `√N`; if none exists, `N` must be prime.**

## 17. Quick Reference

| Python expression | Meaning |
|---|---|
| `num > 1` | Prime numbers must be greater than 1 |
| `num**0.5` | Calculates `√num` |
| `int(num**0.5)` | Takes the integer part of `√num` |
| `range(2, int(num**0.5) + 1)` | Checks divisors from 2 through `√num` |
| `num % i` | Finds the remainder after division |
| `num % i == 0` | `i` divides `num` exactly |
| `break` | Stop once a divisor is found |
| `for-else` | `else` runs if the loop finishes without `break` |
